```python
import requests
import pandas as pd
from datetime import datetime

# Fonksiyon: Belirli tarih aralığı için veri çek
def fetch_weather_data(start_date, end_date, latitude=41.03, longitude=28.84):
    url = f"https://archive-api.open-meteo.com/v1/archive?latitude={latitude}&longitude={longitude}&start_date={start_date}&end_date={end_date}&daily=temperature_2m_max,temperature_2m_min,temperature_2m_mean,precipitation_sum,relative_humidity_2m_mean,windspeed_10m_max,winddirection_10m_dominant,cloudcover_mean&timezone=Europe%2FIstanbul"
    response = requests.get(url)
    if response.status_code == 200:
        data = response.json()['daily']
        df = pd.DataFrame(data)
        df['datetime'] = pd.to_datetime(df['time'])  # Tarih sütunu ekle
        return df
    else:
        print(f"Hata: {response.status_code}")
        return None

# Yıl yıl çek ve birleştir (2012-2025)
years = range(2012, 2026)  # 2025 dahil
all_data = []

for year in years:
    start = f"{year}-01-01"
    end = f"{year}-12-31"
    df_year = fetch_weather_data(start, end)
    if df_year is not None:
        all_data.append(df_year)
    print(f"{year} yılı çekildi.")

# Tüm veriyi birleştir ve CSV'ye kaydet
if all_data:
    df_full = pd.concat(all_data, ignore_index=True)
    df_full.to_csv('istanbul_bagcilar_2012_2025_hava.csv', index=False)
    print("Tam veri 'istanbul_bagcilar_2012_2025_hava.csv' olarak kaydedildi.")
else:
    print("Veri çekilemedi.")
```

```python
import pandas as pd

file_path = '/content/drive/MyDrive/Colab Notebooks/TubitakIoB/CustomerData.csv'

# 1. CSV dosyasını yükle
df = pd.read_csv(file_path)

# 2. Tarih-saat sütunlarını datetime formatına çevir
df['acilis_datetime'] = pd.to_datetime(
    df['ACILIS_TARIHI'].astype(str) + ' ' + df['ACILIS_SAATI'].astype(str),
    errors='coerce'
)

df['kapama_datetime'] = pd.to_datetime(
    df['ACILIS_TARIHI'].astype(str) + ' ' + df['KAPAMA_SAATI'].astype(str),
    errors='coerce'
)

# Oturum süresi hesapla (dakika cinsinden)
df['oturum_sure_dk'] = (df['kapama_datetime'] - df['acilis_datetime']).dt.total_seconds() / 60

# 3. Gruplama: MASANO + CEKNO bazında özetle
df_grup = df.groupby(['MASANO', 'CEKNO', 'acilis_datetime', 'kapama_datetime', 'oturum_sure_dk']).agg(
    toplam_miktar=('MIKTAR', 'sum'),           # Tüm ürünlerin toplam miktarı
    toplam_tutar=('SATIS_FIYATI', 'sum'),      # Oturumun toplam satış tutarı
    urun_listesi=('STOKADI', lambda x: list(x.unique())),  # Oturumdaki benzersiz ürün isimleri (liste)
    ilk_siparis_zamani=('TARIH', 'min')        # Oturumdaki ilk sipariş zamanı (opsiyonel)
).reset_index()

# 4. Hava verisiyle birleştirmek için yuvarlanmış saat ekle
# (Açılış saatini saat başı yuvarla – hava verisi saatlik olduğu için)
df_grup['merge_saati'] = df_grup['acilis_datetime'].dt.floor('H')

# 5. Sonucu yeni bir CSV olarak kaydet
output_path = '/content/drive/MyDrive/Colab Notebooks/TubitakIoB/grup_oturum_bazli_tubitak.csv'
df_grup.to_csv(output_path, index=False, encoding='utf-8-sig')

print("Gruplama tamamlandı!")
print("Yeni dosya: grup_oturum_bazli_tubitak.csv")
print("\nİlk 5 satır örnek:")
print(df_grup.head())
```
